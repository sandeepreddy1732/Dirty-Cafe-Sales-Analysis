```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

df = pd.read_csv("dirty_cafe_sales.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
missing = df.isnull().sum()

missing_percentage = (missing / len(df)) * 100

print(missing_percentage.sort_values(ascending=False))
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
print(df.isnull().sum())
print(df.shape)
```

```python
df.duplicated().sum()
df.info()
```

```python
for col in df.columns:
    print(f"\nColumn: {col}")
    print(df[col].value_counts())
```

```python
df.replace(["ERROR", "UNKNOWN", "Unknown"], np.nan, inplace=True)
```

```python
# Numerical columns
df["Quantity"] = pd.to_numeric(df["Quantity"])
df["Price Per Unit"] = pd.to_numeric(df["Price Per Unit"])
df["Total Spent"] = pd.to_numeric(df["Total Spent"])

# Date column
df["Transaction Date"] = pd.to_datetime(df["Transaction Date"])
```

```python
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDataset Shape:")
print(df.shape)
```

```python
df["Payment Method"] = df["Payment Method"].fillna("Unknown")
df["Location"] = df["Location"].fillna(df["Location"].mode()[0])
```

```python
df.dropna(inplace=True)
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.info())
```

### 1. Top 5 best-selling items

```python
pd.read_sql("select Item, SUM(Quantity) as total_quantity from retail_sales group by item order by total_quantity desc limit 5", con = engine)
```

### 2. Total revenue by payment method

```python
pd.read_sql("""
SELECT `Payment Method`, SUM(`Total Spent`) AS total_revenue
FROM retail_sales
GROUP BY `Payment Method`
ORDER BY total_revenue DESC
""", con=engine)
```

### 3. Top 5 locations by revenue

```python
pd.read_sql("""
SELECT Location, SUM(`Total Spent`) AS total_revenue
FROM retail_sales
GROUP BY Location
ORDER BY total_revenue DESC
LIMIT 5
""", con=engine)
```

### 4. Monthly sales revenue and transaction count

```python
pd.read_sql("""
SELECT
    DATE_FORMAT(`Transaction Date`, '%Y-%m') AS sales_month,
    SUM(`Total Spent`) AS monthly_revenue,
    COUNT(`Transaction ID`) AS total_transactions
FROM retail_sales
GROUP BY sales_month
ORDER BY sales_month
""", con=engine)
```

### 5. Rank items based on total revenue

```python
pd.read_sql("""
SELECT
    Item,
    SUM(`Total Spent`) AS total_revenue,
    DENSE_RANK() OVER (
        ORDER BY SUM(`Total Spent`) DESC
    ) AS revenue_rank
FROM retail_sales
GROUP BY Item
""", con=engine)
```
